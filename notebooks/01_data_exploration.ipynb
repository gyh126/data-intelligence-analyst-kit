# 01. 数据探索 (EDA)

职责1-数据建模：从原始数据中独立完成探索性分析。

**数据源**：火山引擎「专业数据集」vehicle_sales（比亚迪/特斯拉真实销量）+ 多维增强数据。

## 1. 环境与数据加载

```python
import sys; sys.path.insert(0, '../src')
import warnings; warnings.filterwarnings('ignore')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')
%matplotlib inline

from utils.data_loader import prepare_all
ds = prepare_all()
list(ds.keys())
```

## 2. 真实数据概况

```python
byd = ds['byd_monthly']
print('比亚迪月度销量:')
display(byd)
print('\n特斯拉月度销量:')
display(ds['tesla_monthly'])
```

## 3. 销量趋势可视化

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
byd.plot(x='month', y='sales', ax=axes[0], title='比亚迪月度销量', marker='o')
ds['tesla_monthly'].plot(x='month', y='sales', ax=axes[1], title='特斯拉月度销量', marker='o', color='orange')
plt.tight_layout()
```

## 4. 多维度数据分布

```python
md = ds['multi_dim_sales']
print('多维数据 shape:', md.shape)
display(md.head())

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
md.groupby('region')['sales'].sum().plot(kind='bar', ax=axes[0], title='区域销量分布')
md.groupby('channel')['sales'].sum().plot(kind='bar', ax=axes[1], title='渠道销量分布')
md.groupby('model')['sales'].sum().sort_values().plot(kind='barh', ax=axes[2], title='车型销量分布')
plt.tight_layout()
```

## 5. 相关性分析

```python
pivot = md.pivot_table(index='month', columns='region', values='sales', aggfunc='sum')
sns.heatmap(pivot.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('区域间销量相关性')
```

## 6. 日级数据探索

```python
daily = ds['daily_sales']
print(daily.shape)
display(daily.describe())
daily.plot(x='date', y='sales', figsize=(14, 4), title='日级销量时序')
```

## 小结

- 真实数据覆盖比亚迪/特斯拉近12个月销量
- 多维数据含 车型×区域×渠道 三维结构，适合归因分析
- 日级数据含节假日/周末效应，适合时序异常检测
